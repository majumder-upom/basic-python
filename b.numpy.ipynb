Let's demonstrate how `np.random.seed()` works. Notice how the output changes if you comment out or change the seed value.

```
np.random.seed(42)
print('First set of random numbers with seed 42:')
print(np.random.rand(3))

np.random.seed(42) # Set the same seed again
print('\nSecond set of random numbers with seed 42:')
print(np.random.rand(3))

print('\nThird set of random numbers WITHOUT setting seed again:')
print(np.random.rand(3))

np.random.seed(100) # Set a different seed
print('\nFourth set of random numbers with seed 100:')
print(np.random.rand(3))
```
